```python

import os
from dotenv import load_dotenv

load_dotenv("../.env.dev")
os.environ["ANTHROPIC_API_KEY"] = os.getenv("ANTHROPIC_KEY")
from util import *
from llm_functions import *
from run_code_functions import *
import logging

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)
if __name__ == "__main__":
    task = "Implement a Stack data structure with push, pop and peek operations. In Python\n: Pay attention to behave and pytest results"
    output = ""
    try:
        with setup_test_env():
            for iteration in range(1, MAX_RETRIES + 1):
                task+=output
                print(f"\n=== Iteration {iteration} ===")
                result = loop(task)
                
                print("\nSaving implementation files...")
                files, iter_dir = save_test_files(result, iteration)

                print(f"\nGenerated Files in {iter_dir}:")
                for file_type, path in files.items():
                    print(f"{file_type}: {path}")

                print("\nRunning Tests:")
                success, output = run_tests(files)
                print(output)

                with open(f"{iter_dir}/test_results.txt", "w") as f:
                    f.write(output)

                if success:
                    print("\nAll tests passed successfully!")
                    break  # Stop looping if tests pass
                else:
                    print("\nTests failed. Proceeding to next iteration...")

    except Exception as e:
        logger.error(f"Failed to generate/test solution: {e}")
```
